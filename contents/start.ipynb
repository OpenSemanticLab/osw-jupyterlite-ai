# OSW sandbox

This notebook runs entirely in your browser. It talks to the wiki it was served from,
as **you** - there is no login step, and anything it writes is written in your name and
under your rights.

Run the cells in order. The first one takes a little while: the Python runtime and the
OSW packages are being fetched into the browser. It is quicker the next time.

```python
import osw_bootstrap

await osw_bootstrap.setup()
```

```python
# Who does the wiki think we are? A username here rather than an IP means the browser
# session carried into the kernel and nothing else needs configuring.
import js
import requests

info = requests.get(
    f"https://{js.location.hostname}/w/api.php",
    params={"action": "query", "meta": "userinfo", "format": "json"},
).json()
print(info["query"]["userinfo"])
```

```python
# osw-python on top of that same session, with no credentials anywhere.
from osw.wtsite import WtSite

wtsite = osw_bootstrap.wtsite()

page = wtsite.get_page(WtSite.GetPageParam(titles=["Main_Page"])).pages[0]
print(page.title, "-", len(page.get_slot_content("main") or ""), "chars")
```
