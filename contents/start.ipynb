# OSW sandbox

This notebook runs entirely in your browser. It talks to the wiki it was served from,
as **you** - there is no login step, and anything it writes is written in your name and
under your rights.

Run the cells in order.

```python
# Route `requests` through the browser. Without this nothing reaches the wiki at all:
# Pyodide has no sockets, so requests' own transport cannot work.
import pyodide_http

pyodide_http.patch_all()

# The wiki is wherever this notebook came from, so the session cookie applies.
import js

DOMAIN = js.location.hostname
print("wiki:", DOMAIN)
```

```python
# Who does the wiki think we are? If this says a username rather than an IP, the
# browser session carried and nothing else needs configuring.
import requests

info = requests.get(
    f"https://{DOMAIN}/w/api.php",
    params={"action": "query", "meta": "userinfo", "format": "json"},
).json()
print(info["query"]["userinfo"])
```

```python
# osw-python on top of that same session. WtSiteLegacyConfig takes an mwclient.Site we
# built ourselves, which is what lets us skip the credential path entirely.
import mwclient
from osw.wtsite import WtSite

site = mwclient.Site(DOMAIN, path="/w/", scheme="https", do_init=False)
wtsite = WtSite(WtSite.WtSiteLegacyConfig(site=site))

page = wtsite.get_page(WtSite.GetPageParam(titles=["Main_Page"])).pages[0]
print(page.title, "-", len(page.get_slot_content("main") or ""), "chars")
```
